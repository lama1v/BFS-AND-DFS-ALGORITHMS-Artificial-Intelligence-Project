```python
from collections import deque
import random
import time

def bfs(graph, start):
    if start not in graph:
        print(f"Start node {start} not in graph!")
        return set()

    visited = set()
    queue = deque([start])

    print(f"\nBFS starting from {start}:")
    while queue:
        node = queue.popleft()
        if node not in visited:
            print(f"Visited: {node}", end=' -> ')
            visited.add(node)
            for neighbor in graph.get(node, []):
                if neighbor not in visited:
                    queue.append(neighbor)
    print("END")
    return visited
```

```python
def dfs(graph, start):
    if start not in graph:
        print(f"Start node {start} not in graph!")
        return set()

    visited = set()
    stack = [start]

    print(f"\nDFS starting from {start}:")
    while stack:
        node = stack.pop()
        if node not in visited:
            print(f"Visited: {node}", end=' -> ')
            visited.add(node)
            for neighbor in reversed(graph.get(node, [])):
                if neighbor not in visited:
                    stack.append(neighbor)
    print("END")
    return visited
```

```python
directed_graph = {
    'A': ['B', 'C'],
    'B': ['D'],
    'C': ['D'],
    'D': ['E'],
    'E': []
}

undirected_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

print("== Directed Graph ==")
bfs(directed_graph, 'A')
dfs(directed_graph, 'A')

print("\n== Undirected Graph ==")
bfs(undirected_graph, 'A')
dfs(undirected_graph, 'A')
```

```python
def generate_graph(nodes, edges, directed=True):
    if edges > nodes * (nodes - 1) // (1 if directed else 2):
        raise ValueError("Too many edges for the number of nodes")

    graph = {str(i): [] for i in range(nodes)}
    added = set()
    nodes_list = [str(i) for i in range(nodes)]
    random.shuffle(nodes_list)

    for i in range(1, nodes):
        u = nodes_list[i-1]
        v = nodes_list[i]
        graph[u].append(v)
        if not directed:
            graph[v].append(u)
        added.add((u, v))

    while len(added) < edges:
        u, v = random.sample(nodes_list, 2)
        if (u, v) not in added and (not directed or (v, u) not in added):
            graph[u].append(v)
            if not directed:
                graph[v].append(u)
            added.add((u, v))

    return graph

print("\n== Large Graph Test ==")
V, E = 1000, 5000
large_graph = generate_graph(V, E, directed=True)

start = time.time()
bfs(large_graph, '0')
print(f"\nBFS Time: {time.time() - start:.4f} seconds")

start = time.time()
dfs(large_graph, '0')
print(f"DFS Time: {time.time() - start:.4f} seconds")
```
